## 1. Default parameters

Think of ordering chai at a tapri. If you don't say anything, you get normal chai. If you say "kadak", you get kadak. That "normal chai" is the **default**.

A default parameter works the same way: it's the value the function falls back on when you don't give it one.

````python
# 'def' means we are making (defining) a new function.
# sqCalc is just a name we picked. Short for "square calculator".
# n = 1 means: "if nobody gives me n, I'll just take n as 1".
# That 1 is called the default value.
def sqCalc(n = 1):
    # ** means "power of". So n ** 2 is n x n, the square of n.
    # 'return' hands the answer back to whoever called the function.
    return n ** 2
````

````python
# Here we DO pass a value (12), so the default 1 is ignored.
# 12 x 12 = 144
sqCalc(12)
````

````python
# Here we pass nothing. Python sees empty brackets and uses the default, n = 1.
# 1 x 1 = 1
sqCalc()
````

## 2. Keyword arguments

Normally Python matches values by **position**: 1st value goes to the 1st parameter, 2nd to the 2nd, and so on. It doesn't check what the value is, only where it's standing in the line.

**Keyword arguments** let you say exactly which value goes where by writing the parameter's name. Once you do that, the order stops mattering.

````python
# This function needs 3 things, in this exact order: name, age, city.
def printInfo(name,age,city):
    # f"..." is an f-string. Anything inside { } gets swapped with the real value.
    # \n means "jump to a new line", so each detail prints on its own line.
    print(f"Name : {name}\nAge : {age}\nCity : {city}")
````

````python
# Positional arguments: matched by position.
# "ROhan" -> name, 28 -> age, "Gurugram" -> city. Everything lands in the right place.
printInfo("ROhan", 28 , "Gurugram")
````

````python
# Same function, but we swapped the first two values by mistake.
# Python has no idea 28 isn't a name. It just goes: 1st value -> name, 2nd value -> age.
# So the output says "Name : 28" and "Age : ROhan".
# Notice there's NO error. The program happily prints wrong data, which is worse than crashing
# because nobody notices.
printInfo( 28 ,"ROhan", "Gurugram")
````

````python
# Keyword arguments: write the parameter name, then =, then the value.
# city is written first here, but it still goes into city, because we named it.
# Like writing names on tiffin boxes: doesn't matter where you keep them on the table,
# everyone picks up their own.
printInfo(city="Gurugram", name="ROhan",  age=28 )
````

## 3. `*args` (any number of values)

Sometimes you don't know how many values someone will pass. Could be 1, could be 50.

Putting a `*` before a parameter tells Python: **"collect everything they pass and pack it into one tuple."** A tuple is basically a list you can't change after it's made.

People usually call it `*args`, but the name doesn't matter. The `*` is what does the work. Here we called it `*n`.

````python
# *n means "take any number of values and pack them into a tuple called n".
def cube(*n):
    # Print n so we can see what it really looks like inside (it's a tuple).
    print(n)
    # n[0] = the first item in the tuple. Python starts counting from 0, not 1.
    # n[0] ** 3 = that number x itself x itself, i.e. the cube.
    # Heads up: this only cubes the FIRST number. The rest are collected but never used.
    return f"cube of {n[0]} is {n[0] ** 3}"
````

````python
# Only one value passed, so n becomes (12,)
# That comma after 12 is just how Python shows a tuple with a single item.
cube(12)
````

````python
# Five values passed. All five go into n: (122, 234, 45, 2, 3)
# But the function only works on n[0], which is 122. The other four just sit there.
cube(122,234,45,2,3)
````

## 4. `**kwargs` (any number of named values)

`*args` collects loose values into a **tuple**.
`**kwargs` collects **name=value** pairs into a **dictionary**.

A dictionary stores things as pairs: a key (the label) and its value. Like your phone's contact list, where the name is the key and the number is the value.

Easy way to remember: **one star = tuple, two stars = dictionary.**

````python
# **details means "collect every name=value pair into a dictionary called details".
# This new printInfo replaces the old one above (same name, so Python overwrites it).
def printInfo(**details):
    # Print the whole dictionary so we can see how it's stored.
    print(details)
    # details['name'] means "open the dictionary and fetch the value stored under 'name'".
    # Same thing for 'age' and 'city'.
    # Small change from class: we use single quotes inside the { } because the f-string
    # itself uses double quotes. Using the same quotes inside and outside only works on
    # Python 3.12 and newer, so single quotes inside is the safe habit.
    print(f"Name : {details['name']}\nAge : {details['age']}\nCity : { details['city'] }")
````

````python
# We pass four pairs: name, age, city and email.
# All four go into the dictionary, but the function only prints name, age and city.
# email is stored but never used. That's fine, no error.
# BUT if you forget to pass name (or age, or city), you'll get a KeyError,
# because details['name'] won't exist. Try removing name= and see for yourself.
printInfo(name="sanjay" , age = 21 , city="gurugram", email = "sanjay@email.com")
````

## 5. `lambda` (tiny one-line functions)

A lambda is a small function written in a single line, without `def` and without a name of its own. Handy when the job is so small that a full `def` block feels like overkill.

**Format:** `lambda parameters : what to calculate`

Whatever comes after the colon is returned automatically. You don't write `return`.

````python
# The normal (long) way of writing a cube function.
def cb(n):
    # This works out n cubed... and then throws the answer away!
    # There's no 'return', so the function gives back None (Python's word for "nothing").
    # Common mistake. Run the next cell and you'll see.
    n ** 3
````

````python
# Prints None, because cb() calculated the cube but never returned it.
print(cb(2))
````

````python
# The same job as a lambda.
# 'lambda n'  -> this function takes one input called n
# ': n ** 3'  -> work out n cubed and give it back (the return is automatic)
# We store it in the variable cb_calc so we can call it like any normal function.
cb_calc = lambda n : n ** 3
````

````python
# Calling the lambda. 2 x 2 x 2 = 8
cb_calc(2)
````

## 6. List comprehensions

A list comprehension builds a new list in one line, instead of writing a `for` loop with `.append()`.

Read it left to right like a sentence:

`[ what_to_put   for item in collection   if condition ]`

"Give me ___ for every item in ___ (but only if ___)."

````python
# range(1, 11) gives 1, 2, 3 ... 10. The end number (11) is never included.
# 'i for i in range(1,11)' = take each number i and put it in the list as it is.
x = [ i for i in range(1,11)]
````

````python
# Writing just a variable name as the last line of a cell shows its value.
x
````

````python
# Same idea, with a condition at the end.
# i % 2 gives the remainder when i is divided by 2. Remainder 0 means the number is even.
# Read it as: "give me i, for every i in x, but only if i is even".
evens = [ i  for i in x if i % 2 == 0]
````

````python
# Only the even numbers made it in.
evens
````

````python
# x hasn't changed. A comprehension builds a NEW list, it doesn't touch the old one.
x
````

````python
# if/else version. Notice where the if/else sits: when you use else, it goes at the FRONT,
# before the 'for'. (A plain 'if' with no else goes at the END, like the evens example.)
# For each i in x: if it's even make the text "2 even", otherwise make "1 odd".
# We save the result back into x, so the list of numbers gets replaced by a list of text.
# Careful: run this cell only once. On a second run x is full of text like "1 odd",
# and you can't do text % 2, so you'll get an error. Re-run the x = [...] cell above to reset.
x = [f"{i} even" if i % 2 == 0 else f"{i} odd" for i in x ]
````

````python
x
````

````python
# A fresh list of numbers, 1 to 9, for the next examples.
nums = [i for i in range(1,10)]
````

## 7. `filter()` and `map()`

Two jobs come up all the time when working with lists:

- **Picking** only some items (e.g. only the even numbers) -> `filter()`
- **Changing** every item (e.g. squaring each number) -> `map()`

We'll do each job twice: first the long way with a loop, then the short way.

- `filter(function, list)` keeps only the items where the function answers `True`.
- `map(function, list)` runs the function on every item and collects the results.

Both hand back a sort of "packed" object instead of a real list, so we wrap them in `list()` to actually see the values.

````python
# The long way. Start with an empty basket.
evens = []

# Go through the numbers one at a time.
for i in nums:
    # If it divides by 2 with nothing left over, it's even...
    if i % 2 == 0:
        # ...so drop it into the basket.
        evens.append(i)
# Show what ended up in the basket.
evens
````

````python
# Same result using filter().
# The lambda 'x : x % 2 == 0' answers True (even) or False (odd) for each number.
# filter() keeps only the ones that got True.
# list(...) unpacks the result into a normal list we can see.
list(filter( lambda x : x % 2 == 0 , nums) )
````

````python
# The long way of squaring every number. Empty basket again.
sqs = []
for i in nums:
    # Square the number and drop it into the basket.
    sqs.append(i ** 2)
print(sqs)
````

````python
# Same result using map().
# map() takes every number in nums, passes it through the lambda (x ** 2),
# and collects all the answers. list(...) turns them into a normal list.
list(map( lambda x : x ** 2 , nums))
````
