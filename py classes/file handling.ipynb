# File handling (and catching errors)

Every variable you make lives in the computer's short-term memory. The moment the notebook restarts, it's all gone.

**File handling** lets your program write things into a file on your computer and read them back later. Like writing in a notebook instead of just trying to remember.

Every file job follows the same 3 steps:

1. **Open** the file (and tell Python *what* you want to do with it)
2. **Read** or **write**
3. **Close** it

## `open(path, mode)`

- **path** = where the file is. `"./data_file.txt"` means "a file called data_file.txt in the same folder as this notebook". The `./` means "this folder".
- **mode** = what you plan to do with it. This is the important part:

| Mode | What it does | If the file doesn't exist | If the file already exists |
|---|---|---|---|
| `"x"` | create a new file | creates it | **error** (FileExistsError) |
| `"r"` | read only | **error** (FileNotFoundError) | reads it, nothing changes |
| `"w"` | write (overwrite) | creates it | **wipes everything** first, then writes |
| `"a"` | append (add at the end) | creates it | keeps old content, adds new stuff after it |
| `"r+"` | read and write | **error** | doesn't wipe, you can read and write |
| `"w+"` | write and read | creates it | **wipes everything** first, then you can write and read |

If you don't give a mode at all, Python uses `"r"`.

````python
# open(path , mode)

# modes
#  x : create        -> makes a new file, errors if it's already there
#  r : read          -> just look, don't touch (the default)
#  w : overwrite     -> erase everything, then write
#  a : append data   -> add to the end, old stuff stays
#  r+ : read and write   -> file must already exist, nothing gets erased
#  w+ : write and read   -> erases everything first, then lets you write and read
````

## Creating a file with `"x"`

````python
# open() opens the file and gives us a "handle", which is like a remote control for the file.
# We store that remote in f, and use f for everything after this.
# "x" = create a brand new file.
# If data_file.txt is already in the folder, this line throws FileExistsError.
# So running this cell a second time WILL fail. That's normal, the file is already made.
f = open("./data_file.txt", "x")
# write() puts this text inside the file.
f.write("hello world my name is john doe.")
# close() saves the file and lets go of it.
# If you skip this, your text may not actually get saved, and the file can stay "locked".
f.close()
````

## Reading a file: `read()`, `readline()`, `readlines()`

Three ways to read, depending on how much you want:

- `read()` -> the **whole file** as one big string
- `readline()` -> just **one line**
- `readlines()` -> **every line**, as a list (one line per item)

(In class we typed two more lines into the file by hand, that's why the output below shows 3 lines.)

````python
# Open the same file in "r" (read) mode.
f = open("./data_file.txt" , "r")
# read() grabs everything in the file as one string and we store it in data.
data = f.read()
# Show it on screen.
print(data)
# Close it once we're done. (Added this line, it's a good habit to always close.)
f.close()
````

````python
# Open it again. Each open starts reading from the very top of the file.
f = open("./data_file.txt" , "r")
# readline() reads only ONE line: the first one, including the invisible \n at its end.
# That \n, plus print's own new line, is why there's an empty line after the output.
data = f.readline()
print(data)
f.close()
````

````python
f = open("./data_file.txt" , "r")
# readlines() reads every line and gives back a list, one item per line.
# See the \n at the end of each item? That's the hidden "Enter key" that ends a line.
# The last line has no \n, because nobody pressed Enter after typing it.
data = f.readlines()
print(data)
f.close()
````

## Overwriting with `"w"`

**Be careful with this one.** `"w"` doesn't ask, it doesn't warn. The moment you open a file in `"w"` mode, whatever was inside is gone. Like wiping a whiteboard clean before writing on it.

````python
# "w" = write mode. The old 3 lines get erased the instant this line runs.
f = open("./data_file.txt" , "w")
# Now the file contains ONLY this.
f.write("this is the new content")
f.close()
````

````python
# Read it back to prove the old content is really gone.
f = open("./data_file.txt" , "r")
data = f.read()
print(data)
f.close()
````

## Adding more with `"a"`

`"a"` (append) is the safe one. It keeps everything already in the file and adds your new text at the **end**. Like adding a new entry in a diary without tearing out the old pages.

````python
# "a" = append mode. Old content stays safe.
f = open("./data_file.txt" , "a")
# The \n at the start pushes our text onto a fresh line.
# Without it, the new text would get glued to the end of the last line:
# "this is the new contentthis is the new content's second line"
f.write("\nthis is the new content's second line")
f.close()
````

````python
# Read it back. The first line survived, and the new line is added below it.
f = open("./data_file.txt" , "r")
data = f.read()
print(data)
f.close()
````

## Handling errors: `try` / `except` / `else` / `finally`

Things go wrong all the time: the file isn't there, someone divides by zero, a user types "abc" where a number should go. Normally Python just crashes and stops everything.

`try/except` lets you catch the problem and deal with it calmly instead.

- `try` -> "attempt this risky code"
- `except` -> "if it goes wrong, do this instead"
- `else` -> "if it went fine, do this"
- `finally` -> "do this no matter what happened"

````python
# try: attempt the risky code inside.
try:
    # You can't divide by zero, even in maths. Python raises a ZeroDivisionError right here,
    # and jumps straight out of the try block.
    12 / 0
# If the error was a ZeroDivisionError, run this.
# 'as e' saves the error message in e so we can print it.
except ZeroDivisionError as e:
    print("Error :" , e)
# Exception catches pretty much any other kind of error. It's the safety net at the bottom.
# Order matters: Python checks the excepts top to bottom and stops at the first match.
# So always put the specific ones first and the general Exception last.
except Exception as e:
    print("Error", e)
# else runs ONLY if the try block finished with no error at all.
# Here there WAS an error, so this gets skipped.
else:
    print("Task success")
# finally runs no matter what, error or no error.
# Perfect place for cleanup jobs like closing a file.
finally:
    print("Task completed")
````
